```python
import librosa
import pretty_midi
import soundfile
import basic_pitch
import gradio
import numpy as np
```

```python
from google.colab import files
uploaded = files.upload()
```

```python
audio_path = list(uploaded.keys())[0]
y, sr = librosa.load(audio_path, sr=22050)

print("Audio length: ", len(y)/sr)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(14, 5))
plt.plot(y)
plt.xlabel('Time')
plt.ylabel('Amplitude')
plt.title('Waveform')
plt.show()
```

```python
# monophonic pitch detection
f0, voiced_flag, voiced_probs = librosa.pyin(
  y,
  fmin=librosa.note_to_hz('C2'),
  fmax=librosa.note_to_hz('C7'),
  sr=sr
)
```

```python
notes = []

for freq in f0:
    if freq is not None and not np.isnan(freq):
        note = librosa.hz_to_note(freq)
        notes.append(note)

print(notes[:30])
```

```python
compressed = []

prev = None

for n in notes:
    if n != prev:
        compressed.append(n)
        prev = n

print(compressed)
```

```python
target = ['C4', 'E4', 'G4', 'A4']

matches = sum(
    1 for a, b in zip(compressed, target)
    if a == b
)

score = matches / len(target)

print("Accuracy:", score)
```

```python
for i, (played, expected) in enumerate(zip(compressed, target)):
    if played == expected:
        print(f"Note {i+1}: correct")
    else:
        print(f"Note {i+1}: played {played}, expected {expected}")
```
